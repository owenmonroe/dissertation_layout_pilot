## Topic explorer

Look at the pilot BERTopic model (16 topics, 45 issues, frozen extraction v6)
by topic, issue, page or article. Run the first cell, then call any of the
functions below in a new cell.

**Before you start:** use the `topicmodel-env` kernel. Restart the kernel
whenever `topic_explorer.py` changes. Topic names come from `topics.label`.

### The functions

| Function | Use it to | Example |
|---|---|---|
| `explore(t)` | follow one topic: where it is strongest and what sits with it | `explore(9)` |
| `issue(volume, issue_no)` | list any issue's articles in reading order, with each one's topics | `issue("chambers_1843", 575)` |
| `issue_boxes(volume, issue_no)` | read any issue box by box, page by page | `issue_boxes("chambers_1843", 575)` |
| `page(volume, page)` | read any page's topics, column by column | `page("saturday_1834", 22)` |
| `melds(t)` | find articles where topic `t` and another topic share one article | `melds(2)` |
| `article_boxes(doc_id)` | check one article box by box | `article_boxes("chambers_1840/chambers_1840_0018")` |

**Where the arguments come from**
- `t`: a topic number, 0–15 (listed by the first cell).
- `volume`: as in the output folders, e.g. `chambers_1843`, `saturday_1834`,
  `penny`.
- `issue_no`: the printed issue number.
- `page`: the number in the page image's file name
  (`saturday_1834_page_022_order.jpg` → 22).
- `doc_id`: copy it from the `doc_id` column of `issue()`, `page()` or
  `melds()`.

### What `explore(t)` shows
1. **The topic:** its top words and size.
2. **Its strongest issue:** the article carrying it most, then each other
   topic in the issue with its article titles.
3. **Its strongest page beside another article,** in reading order.
4. **Its partner topics across all issues,** with the partner's article
   titles in the issues they share.

Each part can also be run on its own: `overview(t)`, `strongest_issue(t)`,
`strongest_page(t)`, `partners(t)`.

### Meld and juxtaposition

The `relation` column says how a partner topic sits beside topic `t`:
- **same article (meld):** both topics are inside one article;
- **separate article (juxtaposition)** in an issue, or **juxtaposition: same
  column / neighbouring column / same page** on a page: the editor placed
  two articles together.

For melds, use `melds(t)`, then `article_boxes(doc_id)` to see where in the
article each topic falls. Check the boxes before trusting a mix: some are a
real change of subject, some are the model reacting to a few words.

### Options
- `issue(volume, issue_no, min_weight=0.10)`: list an article's other topics only
  at 10% or more (default 5%).
- `strongest_issue(t, rank=1)` and `strongest_page(t, rank=1)`: the
  second-strongest (2 for third, and so on).
- `strongest_page(t, need="topic")`: also counts a partner topic inside the
  topic's own article (meld). `need=None`: the strongest page of all, even
  with nothing else on it.
- `melds(t, other=6)`: only topic `t` with topic 6. `min_share=0.15`: looser
  (default 0.25). `min_words=1000`: longer articles only (default 300).

**Settings** (in `topic_explorer.py`):
- `PARTNER_SHARE = 0.05`: a partner topic counts when it holds at least 5% of
  the issue or page.
- `CARRIES = 0.10`: an article carries a topic when the topic is at least 10%
  of it.

## Melds: one article holding two topics

`melds(t)` lists articles where topic `t` and another topic each hold at least 25% of the article (articles of 300+ words, longest first).
- `melds(2, other=6)`: only topic 2 with topic 6.
- `melds(2, min_share=0.15)`: looser; `min_words=1000`: longer articles only.

Copy a `doc_id` from the table into `article_boxes(...)` to see the article box by box, in reading order, with each box's topic and opening words. Check the boxes before trusting a mix: some mixes are a real change of subject, some are the model reacting to a few words.

```python
melds(1)
```

```python
melds(4)
```

## Any issue

`issue(volume, issue_no)` lists the issue's articles in reading order, one row each: pages, the column it starts in, title, `doc_id`, words, main topic and its share, other topics (5% or more), whether it is a single topic or a meld, series, boxes and flags (short, US reprint, supplement).

`issue_boxes(volume, issue_no)` shows the same issue box by box, with the same columns as `page()`.

`issue_no` is the printed number, e.g. Chambers No. 575 → `issue("chambers_1843", 575)`.

```python
issue("chambers_1843", 575)
```

```python
issue_boxes("chambers_1843", 575)
```
