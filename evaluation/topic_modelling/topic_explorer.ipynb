# Topic explorer

Pick a topic and see where it is strongest and what sits with it, in the
pilot BERTopic model (16 topics, 45 issues, frozen extraction v6).

For the topic, `explore(TOPIC)` shows:
1. **The topic:** its top words and size.
2. **Its strongest issue:**
   - the article carrying it most;
   - every other topic in that issue, with its article titles.

   Each partner is marked **same article (meld)** or **separate article
   (juxtaposition)**.
3. **Its strongest page beside another article,** in reading order. The
   other article holds another topic (at least 5% of the page). Each partner
   article is marked **same column**, **neighbouring column** or **same
   page**: juxtaposition and near-juxtaposition.
4. **Its partner topics across all issues,** with the partner's article
   titles in the issues they share.

**Settings** (in `topic_explorer.py`):
- `PARTNER_SHARE = 0.05`: a partner topic counts when it holds at least 5% of
  the issue or page.
- `CARRIES = 0.10`: an article carries a topic when the topic is at least 10%
  of it.

**Options**
- `strongest_issue(t, rank=1)` and `strongest_page(t, rank=1)` give the
  second-strongest.
- `strongest_page(t, need="topic")` also counts a partner topic inside the
  topic's own article (meld).
- `strongest_page(t, need=None)` gives the strongest page of all, even if it
  holds only that topic.

Run in the `topicmodel-env` kernel. Topic names are suggestions until
`topics.label` is filled in.

## Melds: one article holding two topics

`melds(t)` lists articles where topic `t` and another topic each hold at least 25% of the article (articles of 300+ words, longest first).
- `melds(2, other=6)`: only topic 2 with topic 6.
- `melds(2, min_share=0.15)`: looser; `min_words=1000`: longer articles only.

Copy a `doc_id` from the table into `article_boxes(...)` to see the article box by box, in reading order, with each box's topic and opening words. Check the boxes before trusting a mix: some mixes are a real change of subject, some are the model reacting to a few words.
